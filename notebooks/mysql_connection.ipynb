```python
from pathlib import Path
import pandas as pd
import mysql.connector
```

```python
# load data

FILE_NAME = "demographic.csv"
DATA_PATH = Path("data") / "processed" / FILE_NAME

df = pd.read_csv(DATA_PATH)

df.head()
```

```python
# Create connection
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="BankChurn"
)
```

```python
cursor = conn.cursor()
```

```python
# Push demographic data to database

for index, row in df.iterrows():
    cursor.execute("""
        INSERT INTO demographic (
            CustomerId,
            Gender,
            Age,
            Salary,
            LocationId,
            Churned
        )
        VALUES (%s, %s, %s, %s, %s, %s)
    """,
    (               
    int(row.CustomerId),
    row.Gender,
    int(row.Age),
    float(row.Salary),
    int(row.LocationId),
    int(row.Churned)
    ))
conn.commit()
print("Data inserted successfully")
```
